Problem 2: Write a program that take a user input of three angles and will find out whether it can form a triangle or not.


```python
a=int(input("enter the 1st angle"))
b=int(input("enter the 2nd angle"))
c=int(input("enter the 3rd angle"))

if a+b+c==180:
    print("It can form a triangle")
else:
    print("No")    
```

Problem 3: Write a program that will take user input of cost price and selling price and determines whether its a loss or a profit.


```python
cp=int(input("Enter the cost price "))
sp=int(input("Enter the selling price"))
 
if sp-cp>0:
    print("Profit")
elif sp-cp==0 :
    print("No loss but no profit also")
else:
    print("Loss")       
```

Problem 4: Write a menu-driven program 

cm to ft,
km to miles,
USD to INR,
exit


```python
a=int(input(("""enter the choice
      1. cm to ft
      2. km to miles
      3. USD to INR
      4. Exit""")))


if a==1:
    x=int(input("Enter the cm dimension"))
    ft=x*0.0328
    print("Equivalent feet :", ft)
elif a==2:
    x=int(input("Enter the cm dimension"))
    miles=x*0.621
    print("Equivalent miles :", miles)
elif a==3:
    x=int(input("Enter the cm dimension"))
    miles=x*90
    print("Equivalent INR :", miles)   
elif a==4:
    print("Exiting")
else:
    print("Enter a valid input")         

```

Problem 6 - Find the factorial of a given number.

```python
n=int(input("Enter the number"))
f=1;
for i in range(n,1,-1):
    f=f*i

print(f)


```
